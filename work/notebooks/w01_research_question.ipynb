## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Comparing every page's CTR to one flat threshold is unfair, since CTR varies naturally by position; Discovery B showed this collapse directly. This re-checks that pattern while excluding avg_position == 0 rows, which the data dictionary flags as "no data," not an actual top rank, including them would corrupt any position-based comparison.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

no_position_data = (df["avg_position"] == 0).sum()
print(f"Rows with avg_position == 0 (no data, excluded): {no_position_data}")

visible = df[(df["impressions_90d"] >= 100) & (df["avg_position"] > 0)]
ctr_by_pos = visible.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)
print(ctr_by_pos.round(4).to_string())
```
